```python
import pandas as pd
import numpy as np
from lung_segmentation import segmentation_func,PretrainedUNet
import torch
import os
from tqdm import tqdm
from skimage.transform import resize
```

```python


def load_segmentation_model(model_path):  
    """
        Loading the lung segmentation Model
        
        inputs - 
            model_path: path to the .pt file

        output -
            segmentation_model: lung segmentation model loaded on the cuda
            device: torch.device("cpu") or torch.device("cuda")
    """
    
    device = ("cuda" if torch.cuda.is_available() else "cpu")

    segmentation_model = torch.load(model_path)
    segmentation_model.to(device)
    segmentation_model.eval()
    return (segmentation_model, device)
```

```python
def get_lung_segment(dirname, dest_dirname, segmentation_model, device, image_size):
    filenames = [f for f in os.listdir(dirname) if not f.startswith('.')]
    iter_filenames = tqdm(enumerate(filenames))
    for idx, fname in iter_filenames:
        lung_region = segmentation_func(os.path.join(dirname,fname),segmentation_model,device)
        lung_region = resize(lung_region, (image_size, image_size))
        image_name = fname
        if ".png" in image_name:
            image_name_pt = image_name.replace(".png", "")
        elif ".PNG" in image_name:
            image_name_pt = image_name.replace(".PNG", "")            
        elif ".jpg" in image_name:
            image_name_pt = image_name.replace(".jpg", "")
        elif ".jpeg" in image_name:
            image_name_pt = image_name.replace(".jpeg", "")
        elif ".JPEG" in image_name:
            image_name_pt = image_name.replace(".JPEG", "")    
        elif ".JPG" in image_name:
            image_name_pt = image_name.replace(".JPG", "")    
        np.save(os.path.join(dest_dirname, image_name_pt), lung_region)
        

        

```
